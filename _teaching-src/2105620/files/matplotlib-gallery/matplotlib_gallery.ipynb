# Matplotlib figure gallery

Six **synthetic** Chemical Engineering examples. These are optional extensions to the first-time-user lab. Each example saves an editable SVG, a vector PDF and its synthetic CSV in `gallery_output/`.

Keep this notebook, `matplotlib_gallery.py` and `skh_palette.py` together. The environment needs NumPy, pandas, Matplotlib and Jupyter. Run the setup cell, then an example. Open `matplotlib_gallery.py` to inspect or edit the complete plotting code.


```python
from pathlib import Path
import os
os.environ["MPLCONFIGDIR"] = str(Path.cwd() / ".mpl-cache")
os.environ["XDG_CACHE_HOME"] = str(Path.cwd() / ".cache")
import matplotlib_gallery as gallery
import matplotlib.pyplot as plt
output = Path("gallery_output")
```

## Time-series response with uncertainty

Compare how a response evolves and show between-run spread without hiding the time course.

Synthetic step responses. Lines show the mean of eight independent simulated runs per condition; sand bands show ±1 sample SD at each time. They are not confidence intervals.


```python
fig, metadata = gallery.make_time_response(output_dir=output)
plt.show()
```

## Parity plot with a continuous color variable

Inspect prediction errors and whether they vary with an operating condition. A high correlation alone does not establish agreement.

Synthetic observed and predicted conversion for 52 operating points. Equal axis scales and the graphite identity line reveal agreement. Point color encodes temperature on a continuous ramp; this is an illustration, not model validation.


```python
fig, metadata = gallery.make_parity_temperature(output_dir=output)
plt.show()
```

## Violin, box, and raw observations

```python
fig, metadata = gallery.make_catalyst_distributions(output_dir=output)
plt.show()
```
