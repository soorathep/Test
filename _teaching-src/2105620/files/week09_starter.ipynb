## Variables and lists
Change one value, predict the output, then run the cell.

```python
temperature = [300, 325, 350, 375]
conversion = [40, 55, 70, 80]
print(len(temperature))
print(temperature, conversion)
```

## Reading the runs
Check column names and units before plotting.

```python
df = pd.read_csv("synthetic_reaction_runs.csv")
display(df.head())
print(df.columns.tolist())
print("Runs per temperature:")
print(df.groupby("temperature_C").size())
```

## The first scatter plot
Each mark is one run. Exercise: change the marker size and explain what changed.

```python
fig, ax = plt.subplots(figsize=(5.5, 3.8), layout="constrained")
ax.scatter(df["temperature_C"], df["conversion_pct"],
           color=skh.C["amber"], s=36)
ax.set_xlabel("Temperature (°C)")
ax.set_ylabel("Conversion (%)")
ax.set_ylim(0, 100)
ax.set_xticks(temperature)
plt.show()
```

## Means and sample SD
The sample SD uses n − 1 in the denominator. Confirm that conversion SD is 2 percentage points and selectivity SD is 1 percentage point at each temperature.

```python
summary = df.groupby("temperature_C").agg(
    mean_X=("conversion_pct", "mean"),
    sd_X=("conversion_pct", "std"),
    mean_S=("selectivity_pct", "mean"),
    sd_S=("selectivity_pct", "std")
)
display(summary)
```

## Two panels
Exercise: explain `yerr`, `fmt`, and `capsize`. Then try a narrower y-axis range and decide which range best serves the research question. Keep error-bar definitions in your caption.

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 3.8),
                         sharex=True, layout="constrained")
axes[0].errorbar(summary.index, summary["mean_X"],
                 yerr=summary["sd_X"], fmt="o-",
                 color=skh.C["amber"], capsize=4,
                 linewidth=1.8, markersize=5)
axes[0].set_ylabel("Conversion (%)")
axes[0].set_xlabel("Temperature (°C)")
axes[0].set_ylim(0, 100)
axes[0].set_xticks(temperature)
axes[1].errorbar(summary.index, summary["mean_S"],
                 yerr=summary["sd_S"], fmt="o-",
                 color=skh.C["amber"], capsize=4,
                 linewidth=1.8, markersize=5)
axes[1].set_ylabel("Selectivity to B (%)")
axes[1].set_xlabel("Temperature (°C)")
axes[1].set_ylim(0, 100)
axes[1].set_xticks(temperature)
plt.show()
```
